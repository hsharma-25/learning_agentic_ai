```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel, RunnableSequence, RunnableLambda, RunnablePassthrough
from dotenv import load_dotenv
```

```python
load_dotenv()
model = ChatOpenAI()
parser = StrOutputParser()
```

```python
def countWords(x):
    return len(x.split())
```

```python
prompt1 = PromptTemplate(
    template='Write a joke on {topic}',
    input_variables=['topic']
)

joke_gen_chain = RunnableSequence(prompt1, model, parser)

word_count_chain = RunnableParallel({
    'joke': RunnablePassthrough(),
    'word_count': RunnableLambda(countWords)
})

final_chain = RunnableSequence(joke_gen_chain, word_count_chain)
print(final_chain.invoke({'topic': 'AI'}))
```
