```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
from dotenv import load_dotenv
import requests
```

```python
load_dotenv()
```

```python
##create a tool

@tool
def multiply(a: int, b: int) -> int:
    """given two numbers a and b, this tool returns their product"""
    return a * b
```

```python
llm = ChatOpenAI()
```

```python
##tool binding
llm_with_tools =  llm.bind_tools([multiply])
```

```python
result = llm_with_tools.invoke('How are you?')
result
```

```python
##tool calling

"""LLM doesn't actually run the tool, it just suggests the tool and input arguments"""

messages = []
query = HumanMessage('Can you multiply 3 with 5 ?')
messages.append(query)

result = llm_with_tools.invoke(messages)
```

```python
messages.append(result)
messages
```

```python
result.tool_calls[0]
```

```python
result.tool_calls[0]['args']
```

```python
##tool execution

##sending only the arguments
multiply.invoke(result.tool_calls[0]['args'])
```

```python
##sending the complete tool call
tool_result = multiply.invoke(result.tool_calls[0])
```

```python
messages.append(tool_result)
messages
```

```python
llm_with_tools.invoke(messages)
```
