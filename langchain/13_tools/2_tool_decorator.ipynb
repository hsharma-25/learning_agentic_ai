- custom tool using @tool decorator

```python
from langchain_core.tools import tool
```

```python
##step1: define a function
def multiply(a, b):
    """Multiply two numbers"""
    return (a*b)
```

```python
##step2: add type hints
##type hints help llm to know which inputs to give and what output to expect
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return (a*b)
```

```python
##step3: add tool decorator
##helps llm to communicate with the tool/function
@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return (a*b)
```

```python
result = multiply.invoke({'a': 3, 'b': 5})
result
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```
