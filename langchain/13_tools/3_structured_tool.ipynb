- structured tool in langchain
- enforce strict constraints
- made using pydantic

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description='first number to add')
    b: int = Field(required=True, description='second number to add')
```

```python
def multiply(a, b):
    return (a*b)
```

```python
structured_tool = StructuredTool.from_function(
    func=multiply,
    name='multiply',
    description='multiply two numbers',
    args_schema=MultiplyInput
)
```

```python
result = structured_tool.invoke({'a': 3, 'b': 5})
result
```
