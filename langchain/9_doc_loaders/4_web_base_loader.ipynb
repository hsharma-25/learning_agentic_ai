- extract text from web pages
- uses BeautifulSoup library underneath
- loads only static content
- doesn't handle javascript heavy pages well

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
```

```python
url = 'https://www.croma.com/apple-macbook-air-15-3-inch-m5-16gb-512gb-macos-tahoe-midnight-/p/324345'
loader = WebBaseLoader(url)

doc = loader.load()
```

```python
load_dotenv()
model = ChatOpenAI()
parser = StrOutputParser()
```

```python
prompt = PromptTemplate(
    template='Answer the following question based on the given text: {question} \n {text}',
    input_variables=['question', 'text']
)
```

```python
chain = prompt | model | parser

result = chain.invoke({'question': 'what is the color of the product', 'text': doc[0].page_content})

print(result)
```
