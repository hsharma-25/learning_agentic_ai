```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
import requests
from dotenv import load_dotenv
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_classic import hub
from langchain_classic.agents.agent import AgentExecutor 
from langchain_classic.agents.react.agent import create_react_agent
from langsmith import Client
```

```python
load_dotenv()
llm = ChatOpenAI()
```

```python
search_tool = DuckDuckGoSearchRun()
search_tool.invoke('Top news in India today')
```

```python
@tool
def get_weather_data(city: str) -> str:
    """This function fetches the current weather data for a given city"""

    url = f"http://api.weatherstack.com/current?access_key=b95ecad2a39ff02688824cd039da747a&query={city}"
    response = requests.get(url)

    return response.json()
```

```python
client = Client()

prompt = client.pull_prompt(
    "hwchase17/react",
    dangerously_pull_public_prompt=True
)
```

```python
agent = create_react_agent(
    llm=llm,
    tools=[search_tool, get_weather_data],
    prompt=prompt
)
```

```python
executor = AgentExecutor(
    agent=agent,
    tools=[search_tool, get_weather_data],
    verbose=True
)
```

```python
result = executor.invoke({'input': 'three ways to reach goa from delhi'})
result
```

```python
result = executor.invoke({'input': 'Find the capital of Uttarakhand and get the current weather data for it'})
result
```
